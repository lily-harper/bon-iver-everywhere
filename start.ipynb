# Bon Iver Collaboration Network

This notebook walks through building and visualizing Bon Iver's artist collaboration graph.

**Graph model**
- **Nodes:** artists
- **Edges:** shared song credits (performers, writers, producers)
- **Weights:** number of shared songs between two artists

````python
from boniver.data import load_sample_collaborations, save_graph
from boniver.graph.builder import CollaborationGraphBuilder
from boniver.viz import render_2d, render_interactive_2d, render_interactive_3d

records = load_sample_collaborations()
builder = CollaborationGraphBuilder()
graph = builder.to_networkx(records)

print(builder.graph_summary(graph))
list(graph.edges(data=True))[:5]
````

````python
from pathlib import Path

out_dir = Path("data/output")
save_graph(graph, out_dir / "collaboration_graph.json")

render_2d(graph, out_dir / "collaboration_graph_2d.png")
render_interactive_2d(graph, out_dir / "collaboration_graph_2d.html")
render_interactive_3d(graph, out_dir / "collaboration_graph_3d.html")

print("Open data/output/collaboration_graph_3d.html in your browser.")
````
